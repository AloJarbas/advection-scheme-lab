# Modified-equation follow-up for linear advection
This notebook is the slower companion to the repo's modified-equation sidecar. It asks where the linear diffusion-versus-dispersion story really explains the transport behavior, and where the bounded limiter family breaks away from it.


## Low-wavenumber coefficient definition
For the linear schemes, the one-step amplification factor is expanded as `log G(θ) ≈ -iνθ - D(ν)θ² + iK(ν)θ³`. The card estimates `D` and `K` directly from the exact amplification factor instead of assuming the smooth-wave story speaks for itself.


```python
from advectionlab.analysis import study_modified_equation_followup
rows = study_modified_equation_followup(requested_cfls=(0.2, 0.35, 0.5, 0.65, 0.8, 0.9, 0.95))
[row for row in rows if row.requested_cfl == 0.95 and row.diffusion_coeff is not None]
```

## Where the limiter family leaves the linear curve
At CFL 0.95, Lax-Wendroff is the low-diffusion linear endpoint, but it buys that with visible square-pulse overshoot. The TVD markers matter because they stay on the overshoot floor instead of sitting on the same linear compromise line.


```python
focus_rows = [row for row in rows if row.requested_cfl == 0.95]
[(row.scheme_key, round(row.gaussian_l2_error, 4), round(row.square_l2_error, 4), round(row.square_overshoot, 4)) for row in focus_rows]
```

## Adversarial check
If the bounded family had merely reproduced the same smooth-error versus overshoot curve as the linear schemes, there would be no new lesson here. The useful outcome is that MC and Superbee sit on the zero-overshoot floor while still separating into different smooth and jump preferences.
